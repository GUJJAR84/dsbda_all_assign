```python
len_page=3
```

```python
def reviewsHtml(base_url, total_pages):
    soups = []

    for page in range(1, total_pages + 1):
        if page == 1:
            url = f"{base_url}?sortBy=HELPFUL"
        else:
            url = f"{base_url}?page={page}&sortBy=HELPFUL"

        response = requests.get(url, headers=headers)
        soup = BeautifulSoup(response.text, 'lxml')
        soups.append(soup)

    return soups
```

```python
def getReviews(soup):
    data = []

    reviews = soup.select('div.user-review')

    for r in reviews:
        try:
            name = r.select_one('._reviewUserName')['title'].strip()
        except:
            name = 'N/A'

        try:
            review = r.select_one('p').get_text(strip=True)
        except:
            review = 'N/A'

        try:
            stars = len(r.select('.rating i.sd-icon-star.active'))
        except:
            stars = 'N/A'

        data.append({
            "Name": name,
            "Rating": stars,
            "Review": review
        })

    return data
```

```python
# Step 1 — Fetch all HTML pages
html_datas = reviewsHtml(base_url, len_page)
```

```python
# Step 2 — Extract all reviews
reviews = []

for html_data in html_datas:
    extracted = getReviews(html_data)
    reviews.extend(extracted)   
```

```python
# Step 3 — Convert to DataFrame
df_reviews = pd.DataFrame(reviews)
df_reviews = df_reviews[df_reviews['Name'] != 'N/A']
df_reviews.reset_index(drop=True, inplace=True)

df_reviews
```
